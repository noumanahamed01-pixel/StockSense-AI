```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Load raw data
import pandas as pd
df = pd.read_csv("../data/raw_data.csv", header=[0, 1])

print("Shape:", df.shape)
print("\nColumns:", df.columns.tolist())
df.head()
```

```python
# Flatten the multi-level columns
df.columns = ['_'.join(col).strip() for col in df.columns.values]

print("Columns after fix:", df.columns.tolist())
df.head()
```

```python
# Check actual column names
print("All flattened columns:")
for i, col in enumerate(df.columns):
    print(f"{i}: '{col}'")
```

```python
#Rename columns to simple names
df.rename(columns={
    'Date_Unnamed: 0_level_1' : 'Date',
    'Close_RELIANCE.NS'        : 'Close',
    'High_RELIANCE.NS'         : 'High',
    'Low_RELIANCE.NS'          : 'Low',
    'Open_RELIANCE.NS'         : 'Open',
    'Volume_RELIANCE.NS'       : 'Volume'
}, inplace=True)

print("Renamed columns:", df.columns.tolist())
df.head()
```

```python
df.head()
```

```python
# Convert Date to datetime format
df['Date'] = pd.to_datetime(df['Date'])

# Sort by date (oldest to newest)
df.sort_values('Date', inplace=True)

# Reset index
df.reset_index(drop=True,inplace=True)

print("Date range:", df['Date'].min(), "to", df['Date'].max())
print("Data type of Date:", df['Date'].dtype)
df.head()
```

```python
print("Missing values in each column:")
print(df.isnull().sum())
print(f"\nTotal missing values: {df.isnull().sum().sum()}")
```

```python
# Drop rows where ALL values are missing
df.dropna(how='all', inplace=True)

# Fill any remaining missing values using forward fill
# (use previous day's value to fill gap)
df.ffill(inplace=True)

# Double check
print("Missing values after cleaning:", df.isnull().sum().sum())
print("Shape after cleaning:", df.shape)
```

```python
print("Data types before fix:")
print(df.dtypes)

# Convert price columns to float
price_cols = ['Open', 'High', 'Low', 'Close']
df[price_cols] = df[price_cols].astype(float)

# Convert volume to integer
df['Volume'] = df['Volume'].astype(int)

print("\nData types after fix:")
print(df.dtypes)
```

```python
duplicates = df.duplicated().sum()
print(f"Duplicate rows: {duplicates}")

# Remove if any
df.drop_duplicates(inplace=True)
print(f"Shape after removing duplicates: {df.shape}")
```

```python
print("=" * 45)
print("CLEANED DATA SUMMARY")
print("=" * 45)
print(f"Total rows       : {len(df)}")
print(f"Total columns    : {len(df.columns)}")
print(f"Date range       : {df['Date'].min().date()} to {df['Date'].max().date()}")
print(f"Missing values   : {df.isnull().sum().sum()}")
print(f"Duplicates       : {df.duplicated().sum()}")
print("=" * 45)
df.describe()
```

```python
# Save to processed_data.csv
df.to_csv("../data/processed_data.csv", index=False)
print("✅ Cleaned data saved to data/processed_data.csv")
```
